# Part 1 · What Is a Loop?

A loop tells Python to **repeat** some code, so you don't have to write it again and again.

### 😩 Without a loop

````python
print("Hip hip hooray!")
print("Hip hip hooray!")
print("Hip hip hooray!")
print("Hip hip hooray!")
print("Hip hip hooray!")
````

### 🎉 With a loop

````python
for i in range(5):
    print("Hip hip hooray!")
````

👉 Change the `5` to `20` and run it again. You only had to change **one** number!

🤖 *AI fact: AI models learn by looping over examples millions of times. Loops are how computers practice!*

## Two kinds of loops

| Loop | What it does | Use it when… | In real life |
|---|---|---|---|
| `for` | repeats a **set number of times** | you **know** how many times | "Do 10 jumping jacks." |
| `while` | repeats **as long as something is True** | you **don't know** how many times | "Keep running while you have energy." |

Both loops repeat the **indented** code underneath them. That's called the loop's **body**.

# Part 2 · `for` Loops

## Anatomy of a `for` loop

```python
for i in range(5):
    print("Hello!")
```

| Part | What it does |
|---|---|
| `for` | starts the loop |
| `i` | the **loop variable**: holds the current count (0, then 1, then 2…) |
| `in range(5)` | how many times: **5** |
| `:` | the colon means "the loop body comes next" |
| `····print(...)` | the **4-space indent** puts this line **inside** the loop, so it repeats. Press **Tab** to indent. |

Let's peek inside the loop variable `i`:

````python
for i in range(5):
    print("i is now", i)
````

## Counting with `range()`

| Code | Numbers | Why |
|---|---|---|
| `range(5)` | 0, 1, 2, 3, 4 | starts at **0**, stops **before** 5 |
| `range(1, 6)` | 1, 2, 3, 4, 5 | start at 1, stop before 6 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 | count by **2s** |

**`range(start, stop, step)`: the stop number is never included!**

````python
for n in range(5):
    print(n)
````

````python
for n in range(1, 6):
    print(n)
````

````python
for n in range(0, 10, 2):
    print(n)
````

## Counting backwards and skipping

| Code | Numbers | Why |
|---|---|---|
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 | a **negative** step counts **down** |
| `range(0, 25, 5)` | 0, 5, 10, 15, 20 | count by **5s** |
| `range(10, 0, -2)` | 10, 8, 6, 4, 2 | count down by **2s** |

Counting up or down, the stop number is **never** included.

````python
for n in range(5, 0, -1):
    print(n)
````

````python
for n in range(0, 25, 5):
    print(n)
````

````python
for n in range(10, 0, -2):
    print(n)
````

## Multiplying up with `*=`
`5 × 4 × 3 × 2 × 1` is called **"5 factorial"**. Earlier we added up with `+=`. Now let's **multiply up** with `*=`.

````python
answer = 1
for n in range(1, 6):
    answer *= n
    print("n is", n, "so answer is", answer)
print("5 factorial is", answer)
````

🤔 **Why does `answer` start at 1, not 0?** Anything times 0 is 0! If `answer` started at 0, it would stay 0 forever.

# Part 3 · Comparison Operators

Before we learn `while` loops, we need a new tool. **Comparison operators** compare two values
and answer with a **bool**: `True` or `False`.

| Symbol | Means | Example | Answer |
|:---:|---|:---:|:---:|
| `>` | greater than | `7 > 2` | `True` |
| `<` | less than | `7 < 2` | `False` |
| `>=` | greater than or equal | `5 >= 5` | `True` |
| `<=` | less than or equal | `3 <= 1` | `False` |
| `==` | equal to | `7 == 7` | `True` |
| `!=` | not equal to | `7 != 2` | `True` |

````python
print(7 > 2)
print(7 < 2)
print(5 >= 5)
print(3 <= 1)
print(7 == 7)
print(7 != 2)
````

### `=` vs `==`
- **One** `=` **puts** a value in the box.
- **Two** `==` **asks** a question: "are these equal?"

````python
age = 12          # put 12 in the box
print(age == 12)  # is age equal to 12?
print(age == 13)  # is age equal to 13?
````

Comparisons work great with variables:

````python
height = 135
print("Tall enough for the roller coaster?", height >= 120)

lives = 0
print("Game over?", lives == 0)
````

# Part 4 · `while` Loops

A `while` loop keeps repeating **as long as** its condition is `True`. It checks the question **before every round**.

````python
fuel = 3
while fuel > 0:
    print("Vroom! Fuel:", fuel)
    fuel -= 1
print("Out of fuel!")
````

| Part | What it does |
|---|---|
| `while` | starts the loop |
| `fuel > 0` | the **condition**: a True/False question. While it's `True`, keep going. |
| `fuel -= 1` | changes the variable each round, so the loop can **end** |

## How a `while` loop thinks

| `fuel` | `fuel > 0`? | What happens |
|---|---|---|
| 3 | `True` | print 3, fuel → 2 |
| 2 | `True` | print 2, fuel → 1 |
| 1 | `True` | print 1, fuel → 0 |
| 0 | `False` | leave the loop! |

👉 Change `fuel = 3` to `fuel = 6` and run it again. What if you start with `fuel = 0`?

## ⚠️ Watch out: infinite loops!
If nothing inside the loop changes the condition, it stays `True` **forever**:

```python
fuel = 3
while fuel > 0:
    print("Vroom! Fuel:", fuel)
    # oops! we forgot fuel -= 1
```

`fuel` is always 3, so `3 > 0` is always `True`, and the loop never stops!

## The magic doubling penny 🪙
You get **1 penny** today. Every day, it **doubles**. How many days until you have **1,000,000 pennies**?

🤔 **Guess first!** 100 days? 1,000 days? Write your guess here: ______

We **don't know** how many days it will take, so we use a `while` loop and keep doubling **until** we reach the goal.

````python
pennies = 1
days = 0

while pennies < 1000000:
    pennies *= 2
    days += 1

print("Days:", days)
print("Pennies:", pennies)
````

Were you close? 👉 Add `print(days, pennies)` **inside** the loop to watch the pennies grow each day.

# 🏁 What we learned today

- **`for` loops** repeat a set number of times. `range()` does the counting.
- **`range(start, stop, step)`**: the stop is never included, and a negative step counts down.
- **Build it up:** use `+=` or `*=` inside a loop to add up or multiply up. Multiply from 1!
- **Comparisons** `> < >= <= == !=` ask questions and answer `True` or `False`.
- **`while` loops** repeat as long as a condition is `True`. They're great when you don't know how many times.
- **Infinite loops:** something inside the loop must change, or it never stops! Use **■ Stop** to escape.
