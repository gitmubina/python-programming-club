# Part 1 · Recap: Loops

## `for` loops: repeat a set number of times
Don't forget the **`:`** and the **4-space indent**!

````python
for i in range(5):     # repeat 5 times
    print("Hello!")    # indented = inside the loop
print("Done!")         # not indented = runs once
````

| Code | Numbers |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(10, 0, -1)` | 10, 9, 8 … 1 |

The stop number is **never** included. Use `total += n` to **add up** and `answer *= n` to **multiply up** (start at 1!).

## `while` loops: repeat as long as something is True

````python
fuel = 3
while fuel > 0:          # the condition
    print("Vroom!", fuel)
    fuel -= 1            # change it, so the loop can end!
print("Out of fuel!")
````

**Comparison operators** `>  <  >=  <=  ==  !=` ask questions and answer `True` or `False`.

⚠️ If nothing inside a `while` loop changes the condition, it runs **forever**. Press **■ Stop** to escape!

# Part 2 · `if` and `else`

## Making decisions
You make decisions all day long. Programs do too!
- ☔ **If** it's raining, take an umbrella.
- 🏆 **If** your score is 100 or more, you win!
- 🚦 **If** the light is green, go.

The **condition** is a True/False question, just like in `while` loops. If it's `True`, the indented code runs.
If it's `False`, Python skips it.

````python
raining = True

if raining:
    print("Take an umbrella! ☔")
````

👉 Change `raining` to `False` and run it again. What happens?

## Anatomy of an `if` statement

```python
if score >= 100:
    print("You win!")
```

| Part | What it does |
|---|---|
| `if` | starts the decision |
| `score >= 100` | the **condition**: a True/False question |
| `:` | the colon means "the code to run comes next" |
| `····print(...)` | the **4-space indent** puts this line **inside** the if. It only runs when the condition is `True`. |

💡 **Sound familiar?** It looks just like a `while` loop! The difference: `if` checks **once**, `while` keeps repeating.

````python
score = 120

if score >= 100:
    print("You win! 🏆")
print("Thanks for playing!")
````

👉 Change `score` to `80`. Which line still prints? Why?

## `if` / `else`: one or the other
`else` means **"otherwise"**. It catches everything the `if` didn't. **Exactly one** of the two blocks runs, never both.

````python
raining = True

if raining:
    print("Take an umbrella! ☔")
else:
    print("Wear sunglasses! 😎")
````

# Part 3 · More Choices with `elif`

`elif` means **"else if"**. Python checks from **top to bottom**, runs the **first** one that's `True`,
then skips the rest.

````python
score = 85

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("Keep practicing!")
````

| Check | Answer | What happens |
|---|---|---|
| `score >= 90` | `False` | skip |
| `score >= 80` | `True` | print `"B"` ✅ |
| `score >= 70` | – | skipped |
| `else` | – | skipped |

👉 `85` is **also** bigger than `70`, so why doesn't it print `C` too? Try `score = 95`, `75` and `40`.

# Part 4 · Combining Questions: `and`, `or`, `not`

| Word | Means | Example |
|---|---|---|
| `and` | **both** must be True | `age >= 10 and height >= 130` |
| `or` | **at least one** must be True | `day == "Sat" or day == "Sun"` |
| `not` | **flips** True ↔ False | `not raining` |

````python
age = 12
height = 140

if age >= 10 and height >= 130:
    print("You can ride! 🎢")
else:
    print("Sorry, not this time.")
````

````python
raining = False

if not raining:
    print("Let's play outside! ⚽")
````

# Part 5 · Decisions Inside Loops

Put an `if` **inside** a loop and Python makes a new decision **every round**.
Work out what each cell prints **before** you run it!

### Even or odd?
Remember the `%` trick? `n % 2 == 0` means `n` is even.

````python
for n in range(1, 6):
    if n % 2 == 0:
        print(n, "is even")
    else:
        print(n, "is odd")
````

### Count the 3s
How many numbers from 1 to 20 can be shared evenly by 3?

````python
count = 0
for n in range(1, 21):
    if n % 3 == 0:
        count += 1
print(count, "numbers")
````

### Health check
The `if` only prints when health drops below 50.

````python
health = 100
while health > 0:
    health -= 40
    if health < 50:
        print("Low!", health)
print("Game over")
````

# 🏁 What we learned today

- **`if` / `else`**: run code only when a condition is `True`. `else` catches everything else.
- **`elif`**: more than two choices. Python runs the **first** True one and skips the rest, so order matters!
- **`and` / `or` / `not`**: combine questions: both True, at least one True, or flip it.
- **Colon & indent**: every `if`, `elif` and `else` line ends with `:`, and the code inside is indented.
- **Decisions in loops**: put an `if` inside a loop and Python decides again every round.
