# Part 2 · Arithmetic Operators

## Our math toolbox

| Operator | Name | Example | Result |
|:---:|---|:---:|:---:|
| `+` | Addition | `7 + 2` | `9` |
| `-` | Subtraction | `7 - 2` | `5` |
| `*` | Multiplication | `7 * 2` | `14` |
| `/` | Division | `7 / 2` | `3.5` |
| `//` | Floor division (whole part only) | `7 // 2` | `3` |
| `%` | Modulus (the remainder) | `7 % 2` | `1` |
| `**` | Exponent (power) | `7 ** 2` | `49` |

````python
a = 7
b = 2

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

💡 **`//` and `%` are a team.** If you share 17 cookies with 5 friends, `//` tells you how many each friend gets,
and `%` tells you how many are left over.

````python
cookies = 17
friends = 5

print("Each friend gets:", cookies // friends)
print("Left over:", cookies % friends)
````

## Rules to remember

### Order matters (PEMDAS)
**P**arentheses → **E**xponents → **M**ultiply and **D**ivide → **A**dd and **S**ubtract. When in doubt, add parentheses!

````python
print(2 + 3 * 4)     # multiply first
print((2 + 3) * 4)   # parentheses first
````

### Division makes floats
`/` always gives a decimal. `//` gives a whole number.

````python
print(10 / 2)
print(10 // 2)
````

### Update shortcuts
| Shortcut | Same as |
|---|---|
| `score += 5` | `score = score + 5` |
| `score -= 5` | `score = score - 5` |
| `score *= 5` | `score = score * 5` |

````python
score = 0
score += 5
print(score)
score *= 3
print(score)
score -= 1
print(score)
````

### Works on text too
`+` glues strings together. `*` repeats them.

````python
print("ha" * 3)
print("Py" + "thon")
````

# Part 3 · `for` Loops

## What is a loop?
A loop tells Python to **repeat** some code, so you don't have to write it again and again.

### 😩 Without a loop

````python
print("Hip hip hooray!")
print("Hip hip hooray!")
print("Hip hip hooray!")
print("Hip hip hooray!")
print("Hip hip hooray!")
````

### 🎉 With a loop

````python
for i in range(5):
    print("Hip hip hooray!")
````

👉 Change the `5` to `20` and run it again. You only had to change **one** number!

🤖 *AI fact: AI models learn by looping over examples millions of times. Loops are how computers practice!*

## Anatomy of a `for` loop

```python
for i in range(5):
    print("Hello!")
```

| Part | What it does |
|---|---|
| `for` | starts the loop |
| `i` | the **loop variable**: holds the current count (0, then 1, then 2…) |
| `in range(5)` | how many times: **5** |
| `:` | the colon means "the loop body comes next" |
| `····print(...)` | the **4-space indent** puts this line **inside** the loop, so it repeats. Press **Tab** to indent. |

Let's peek inside the loop variable `i`:

````python
for i in range(5):
    print("i is now", i)
````

### Forgetting the colon or the indent
> ⚠️ **The next cell shows an error on purpose.** Read the last line of the red message. Python is telling you what went wrong.

## Counting with `range()`

| Code | Numbers | Why |
|---|---|---|
| `range(5)` | 0, 1, 2, 3, 4 | starts at **0**, stops **before** 5 |
| `range(1, 6)` | 1, 2, 3, 4, 5 | start at 1, stop before 6 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 | count by **2s** |

**`range(start, stop, step)`: the stop number is never included!**

````python
for n in range(5):
    print(n)
````

````python
for n in range(1, 6):
    print(n)
````

````python
for n in range(0, 10, 2):
    print(n)
````

# Part 4 · Comparison Operators

Before we learn `while` loops, we need a new tool. **Comparison operators** compare two values
and answer with a **bool**: `True` or `False`.

| Symbol | Means | Example | Answer |
|:---:|---|:---:|:---:|
| `>` | greater than | `7 > 2` | `True` |
| `<` | less than | `7 < 2` | `False` |
| `>=` | greater than or equal | `5 >= 5` | `True` |
| `<=` | less than or equal | `3 <= 1` | `False` |
| `==` | equal to | `7 == 7` | `True` |
| `!=` | not equal to | `7 != 2` | `True` |

````python
print(7 > 2)
print(7 < 2)
print(5 >= 5)
print(3 <= 1)
print(7 == 7)
print(7 != 2)
````

### `=` vs `==`
- **One** `=` **puts** a value in the box.
- **Two** `==` **asks** a question: "are these equal?"

````python
age = 12          # put 12 in the box
print(age == 12)  # is age equal to 12?
print(age == 13)  # is age equal to 13?
````

Comparisons work great with variables:

````python
height = 135
print("Tall enough for the roller coaster?", height >= 120)

lives = 0
print("Game over?", lives == 0)
````

# Part 5 · `while` Loops

A `while` loop keeps repeating **as long as** its condition is `True`. It checks the question **before every round**.

````python
fuel = 3
while fuel > 0:
    print("Vroom! Fuel:", fuel)
    fuel -= 1
print("Out of fuel!")
````

| Part | What it does |
|---|---|
| `while` | starts the loop |
| `fuel > 0` | the **condition**: a True/False question. While it's `True`, keep going. |
| `fuel -= 1` | changes the variable each round, so the loop can **end** |

## How a `while` loop thinks

| `fuel` | `fuel > 0`? | What happens |
|---|---|---|
| 3 | `True` | print 3, fuel → 2 |
| 2 | `True` | print 2, fuel → 1 |
| 1 | `True` | print 1, fuel → 0 |
| 0 | `False` | leave the loop! |

👉 Change `fuel = 3` to `fuel = 6` and run it again. What if you start with `fuel = 0`?

## ⚠️ Watch out: infinite loops!
If nothing inside the loop changes the condition, it stays `True` **forever**:

```python
fuel = 3
while fuel > 0:
    print("Vroom! Fuel:", fuel)
    # oops! we forgot fuel -= 1
```

`fuel` is always 3, so `3 > 0` is always `True`, and the loop never stops!

# 🏁 What we learned today

- **Operators** `+ - * / // % **` do math. `//` gives the whole part, and `%` gives the remainder.
- **Rules:** PEMDAS order, `/` always makes a float, and `+=` is a shortcut for updating.
- **Comparisons** `> < >= <= == !=` ask questions and answer `True` or `False`.
- **`for` loops** repeat a set number of times. `range()` does the counting, and the stop number is never included.
- **`while` loops** repeat as long as a condition is `True`.
- **Infinite loops:** something inside the loop must change, or it never stops! Use **■ Stop** to escape.
