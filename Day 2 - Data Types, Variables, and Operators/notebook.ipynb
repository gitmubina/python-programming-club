# Part 1 · Data Types

## What are data types?

A **data type** tells Python **what kind of information** a value is, and **what it can do with it**.

Look at these two cells. They use the **same symbols** but give **different answers**. Run them both.

````python
# numbers get added
5 + 5
````

````python
# text gets glued together
"5" + "5"
````

🤔 **What happened?** The quotes changed the type. Without quotes, `5` is a **number**, so `+` adds.
With quotes, `"5"` is **text**, so `+` glues the pieces together.

## The four basic data types in Python

| Type | Full name | What it holds | Examples |
|------|-----------|---------------|----------|
| `int` | Integer | Whole numbers, positive or negative | `7`  `-3`  `100` |
| `float` | Float | Numbers with a decimal point | `3.14`  `-0.5`  `2.0` |
| `str` | String | Text, **always inside quotes** | `"hi"`  `"Maya"`  `"42"` |
| `bool` | Boolean | Only two values: yes or no | `True`  `False` |

### Not sure what type something is? Ask Python!
`type()` tells you the data type of any value.

````python
type(3.5)
````

### `int`: whole numbers

````python
print(type(7))
print(type(-3))
print(type(100))
````

### `float`: numbers with a decimal point

````python
print(type(3.14))
print(type(-0.5))
print(type(2.0))   # still a float! The decimal point is what counts
````

### `str`: text inside quotes

````python
print(type("hi"))
print(type("Maya"))
print(type("42"))    # looks like a number, but the quotes make it text
print(type('hello')) # single quotes work too
````

### `bool`: True or False

````python
print(type(True))
print(type(False))
````

💡 `True` and `False` **must start with a capital letter**. Python does not know what `true` means.

# Part 2 · Variables

## What are variables?

A **variable** is a **name that stores a value**, like a **labeled box** you can open later.

| Box label (variable) | What's inside (value) | Python code |
|---|---|---|
| `name` | `"Maya"` | `name = "Maya"` |
| `age` | `12` | `age = 12` |
| `game_over` | `False` | `game_over = False` |

### ⚠️ `=` does NOT mean "equals"
In Python, `=` means **"put this value in the box."** It is not the "equals" you know from math class.

````python
name = "Maya"
age = 12
game_over = False
````

🤔 **Nothing printed?** That's normal. Python stored the values quietly. To look inside the boxes, use `print()`:

````python
print(name)
print(age)
print(game_over)
````

A variable has the same type as the value stored inside it:

````python
print(type(name))
print(type(age))
print(type(game_over))
````

## How do we use variables?

There are three steps: **Create → Use → Update**.

````python
# 1 · CREATE: makes the box and puts 0 in it
score = 0

# 2 · USE: looks inside the box
print(score)
````

````python
# 3 · UPDATE: replaces the old value
score = score + 10
print(score)
````

🧠 **How to read `score = score + 10`:**
1. Python works out the **right side** first: *what's in `score` right now, plus 10.*
2. Then it puts the answer back into the box on the **left**.

👉 Try running the cell above **2 or 3 more times**. What happens to the score each time?

A new value **replaces** the old one. A box holds only one value at a time.

````python
lives = 3
print(lives)

lives = 2      # the old value (3) is thrown away
print(lives)
````

## Naming your variables

| ✅ Good names | ❌ Not allowed | Why not? |
|---|---|---|
| `player_name` | `2lives` | starts with a number |
| `high_score` | `my score` | has a space |
| `lives2` | `high-score` | has a dash |

**Rules:** use letters, numbers and underscores `_`. Don't start with a number.
**Tip:** pick a name that describes what's inside. `high_score` is much clearer than `x`.

````python
player_name = "Sam"
high_score = 950
lives2 = 3

print(player_name)
print(high_score)
print(lives2)
````

# Part 3 · How Variables Work Together

Variables can be combined to make **new** values. What you can do depends on their **data type**.

## Numbers: do math

````python
price = 3
quantity = 4
total = price * quantity
print(total)
````

## Strings: glue together

````python
first = "Ada"
last = "Lovelace"
full = first + " " + last
print(full)
````

💡 *Fun fact: Ada Lovelace wrote what many people call the **first computer program**, back in the 1840s!*

🤔 Why did we add `" "` in the middle? Run this cell and see what happens without it:

````python
print(first + last)
````

## ⚠️ Careful: mixing types
Python **won't** add text and a number together. It doesn't know whether you want to *add* or to *glue*.

> ⚠️ **The next cell shows an error on purpose.** Read the last line of the red message. Python is telling you what went wrong.

````python
"Age: " + 12
````

**The fix:** `str()` turns a number into text, so Python can glue it.

````python
"Age: " + str(12)
````

````python
age = 12
message = "I am " + str(age) + " years old"
print(message)
````

**Going the other way:** `int()` and `float()` turn text into numbers.

````python
text_number = "42"

print(text_number + "8")       # both are text, so they get glued
print(int(text_number) + 8)    # int() makes it a number, so they get added
print(float("2.5") + 1)
````

💡 **Shortcut:** remember the commas from Part 0? `print()` can show several things at once if you separate them with commas.
It handles mixed types for you and adds the spaces too.

````python
age = 12
print("Age:", age)
````

## Why do we need variables?

Without variables, a program would **forget everything** the moment it happened.

| 🧠 They remember | 🔁 Change once | 📖 Easy to read |
|---|---|---|
| A game keeps track of your score, lives and level while you play. | Update `price` in one place and every calculation that uses it updates too. | `price * quantity` explains itself. `3 * 4` doesn't. |

### 🧠 They remember

````python
score = 0
lives = 3

# you collect a coin
score = score + 10

# you collect another coin
score = score + 10

# you bump into a monster
lives = lives - 1

print("Score:", score)
print("Lives:", lives)
````

# Part 4 · Arithmetic Operators

## What are arithmetic operators?

**Symbols that tell Python to do math.** Here's each one with the same two numbers, **7** and **2**:

| Operator | Name | Example | Result |
|:---:|---|:---:|:---:|
| `+` | Addition | `7 + 2` | `9` |
| `-` | Subtraction | `7 - 2` | `5` |
| `*` | Multiplication | `7 * 2` | `14` |
| `/` | Division | `7 / 2` | `3.5` |
| `//` | Floor division (whole part only) | `7 // 2` | `3` |
| `%` | Modulus (the remainder) | `7 % 2` | `1` |
| `**` | Exponent (power) | `7 ** 2` | `49` |

````python
a = 7
b = 2

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

### `//` and `%`: sharing cookies 🍪
You have **17 cookies** and **5 friends**.
- `//` tells you how many **whole** cookies each friend gets.
- `%` tells you how many are **left over**.

````python
cookies = 17
friends = 5

print("Each friend gets:", cookies // friends)
print("Left over:", cookies % friends)
````

💡 **`%` trick: even or odd?** If `number % 2` is `0`, the number is **even**. If it's `1`, the number is **odd**.

````python
print(10 % 2)
print(7 % 2)
````

### `**`: powers
`2 ** 3` means `2 * 2 * 2`.

````python
print(2 ** 3)
print(10 ** 6)    # a million!
print(2 ** 100)   # Python can handle really big numbers
````

## How do we use operators?

### Order matters: PEMDAS
Python follows the same rules as math class:
**P**arentheses → **E**xponents → **M**ultiply and **D**ivide → **A**dd and **S**ubtract.

When in doubt, add parentheses!

````python
print(2 + 3 * 4)     # multiply first: 2 + 12
print((2 + 3) * 4)   # parentheses first: 5 * 4
````

### Division makes floats
`/` **always** gives a decimal (a `float`), even when it divides evenly. Use `//` for a whole number.

````python
print(10 / 2)
print(10 // 2)

print(type(10 / 2))
print(type(10 // 2))
````

If you mix an `int` and a `float`, the answer is a `float`:

````python
print(3 + 2.0)
print(type(3 + 2.0))
````

### 💥 You can't divide by zero
Not in math class, and not in Python either!

> ⚠️ **The next cell shows an error on purpose.** Read the last line of the red message. Python is telling you what went wrong.

````python
print(5 / 0)
````

### Update shortcuts
A quicker way to change a variable's value.

| Shortcut | Same as |
|---|---|
| `score += 5` | `score = score + 5` |
| `score -= 5` | `score = score - 5` |
| `score *= 5` | `score = score * 5` |
| `score /= 5` | `score = score / 5` |

````python
score = 0

score += 5     # same as score = score + 5
print(score)

score -= 2
print(score)

score *= 10
print(score)
````

### Works on text too!
`+` glues strings together. `*` repeats them.

````python
print("ha" * 3)
print("Py" + "thon")
print("-" * 20)
print("Go" + "!" * 5)
````
